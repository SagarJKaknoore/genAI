```python
from langchain.agents import create_agent
from langchain_community.tools import tool
from langchain_ollama import ChatOllama


@tool
def get_weather(city: str) -> str:
    """Get weather for a city."""
    return f"The weather in the city {city} is sunny"


@tool
def get_stock(stock: str) -> str:
    """Get the current stock price for a company."""
    return f"The current stock price of {stock} is ₹1000"


model = ChatOllama(
    model="llama3.2:latest"
)


ALL_TOOLS = [get_weather, get_stock]


agent = create_agent(
    model=model,
    tools=ALL_TOOLS,
    system_prompt="""
    You are a helpful assistant.

    Use the get_weather tool when the user asks about weather.
    Use the get_stock tool when the user asks about stock prices.
    """
)


response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is today stock"
        }
    ]
})


print(response)
```
