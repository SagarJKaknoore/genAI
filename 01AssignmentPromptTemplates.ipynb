```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2:latest")
print("Model Ready")
```

```python
from langchain_core.prompts import PromptTemplate
user_prompt_template = PromptTemplate(
    template= "Please give details of the touriest place {touriest_place} located in the city {city} of the country {country} "
)
user_prompt =user_prompt_template.format(touriest_place="Mysore Palace", city="Mysore", country="India")
response = model.invoke(user_prompt)
print(type(response))
print(response)                           
```

```python
from langchain_core.prompts import ChatPromptTemplate
user_prompt_template = ChatPromptTemplate([("user","Please give details of the touriest place {touriest_place} located in the city {city} of the country {country} "
)])
user_prompt =user_prompt_template.format(touriest_place="Mysore Palace", city="Mysore", country="India")
response = model.invoke(user_prompt)
print(type(response))
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate
user_prompt_template = ChatPromptTemplate([("user","Please give details of the touriest place {touriest_place} located in the city {city} of the country {country} "
), ("system","You are a travel expert")])
user_prompt =user_prompt_template.format(touriest_place="Mysore Palace", city="Mysore", country="India")
response = model.invoke(user_prompt)
print(type(response))
print(response)
```
