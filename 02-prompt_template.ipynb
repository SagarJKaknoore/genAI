```python
from langchain_ollama import ChatOllama

# 1. Create LLM
llm = ChatOllama(model="llama3.2:latest")
```

```python
from langchain_core.prompts import PromptTemplate


# 2. Create Prompt Template
prompt = PromptTemplate(
    template="What is the capital city of {country}?",
    input_variables=["country"]
)

# 3. Fill the template
user_prompt = prompt.format(country="France")

# 4. Send prompt to LLM
response = llm.invoke(user_prompt)

# 5. Print response
print(type(response))
print(response)
```

```python
from langchain_core.prompts import PromptTemplate

# Input data
input_data = {
    "name": "Sachin Tendulkar",
    "team": "Indian Cricket Team",
}

# Create PromptTemplate with Mustache variables
user_prompt_template = PromptTemplate(
    template="{{name}} is a famous cricketer who played for {{team}}.",
    input_variables=["name", "team"],
    template_format="mustache"
)

# Fill the template
user_prompt = user_prompt_template.format(**input_data)

# Send to LLM
response = llm.invoke(user_prompt)

# Print response
print(type(response))
print(response)
```
