```python
from langchain_core.prompts import PromptTemplate

one_shot_prompt_template = """
Example:
Question: I have a meeting with my manager.
Answer: Work

Now answer this:
Question: {question}
Answer:
"""
user_prompt_template = PromptTemplate(
    template=one_shot_prompt_template,
    input_variables=["question"]
)

user_template=user_prompt_template.format(question="I am going to mall to By a shirt")
response=llm.invoke(user_template)
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
question = "I am going to the mall to enjoy part with girl fiend"
one_shot_prompt_template = (
    f"Example:\n"
    f"Question: I have a meeting with my manager.\n"
    f"Answer: Work\n\n"
    f"Now answer this:\n"
    f"Question: {question}\n"
    f"Answer:"
)

user_prompt_template = PromptTemplate(
    template=one_shot_prompt_template,
    input_variables=["question"]
)

user_template=user_prompt_template.format(question=question)
response=llm.invoke(user_template)
print(response.content)
```
