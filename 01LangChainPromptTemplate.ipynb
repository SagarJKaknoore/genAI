```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2:latest")
print("Model Ready")
```

```python
from langchain_core.prompts import PromptTemplate

user_prompt_template = PromptTemplate(
    template="Who is the chief minister of {state}?",
    input_variables=["state"]
)

user_prompt = user_prompt_template.format(state="Karnataka")

response = model.invoke(user_prompt)

print(type(response))
print(response)
```

```python
#Example of using prompt template with mustache variables

from langchain_core.prompts import PromptTemplate

input = {
    "state": "Karnataka",
    "topic": "Sandalwood"
}

user_prompt_template= PromptTemplate(
    template ="who is the famous actor in {topic} ,and who is the chief minister of {state}?",
    input_variables=["state","topic"]
)

user_prompt= user_prompt_template.format(**input)
response = model.invoke(user_prompt)
print(type(response))
print(response)
```
