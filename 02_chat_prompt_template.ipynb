```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2:latest")
print("Model is Ready")
```

```python
from langchain_core.prompts import ChatPromptTemplate

# Create Chat Prompt Template
user_chat_prompt_template = ChatPromptTemplate([
    ("user", "What is the capital city of {country}?")
])

# Provide value for the variable
user_prompt = user_chat_prompt_template.invoke({
    "country": "France"
})

# Send the formatted chat prompt to the LLM
response = llm.invoke(user_prompt)

# Print response
print(type(response))
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate

# Create Chat Prompt Template
user_chat_prompt_template = ChatPromptTemplate([
    ("user", "Please explain the {concept} in simple terms."),
    ("system", "You are illiterate Person who cannot read or write, Please explain the {concept} in simple terms that I can understand.")
])

# Provide value for the variable
user_prompt = user_chat_prompt_template.invoke({
    "concept": "pythagorean theorem"
})

# Send the formatted chat prompt to the LLM
response = llm.invoke(user_prompt)

# Print response
print(type(response))
print(response)
```

```python
#Explaination of Conversation state
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

user_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{user_input}",),
]
)
chat_history =[
    HumanMessage(content="Which is the capital city of France?"),
    AIMessage(content="The capital city of France is Paris."),
]
user_prompt = user_prompt_template.format_messages(chat_history=chat_history, user_input="Which is famous landmark in there?")

response = llm.invoke(user_prompt)

print(type(response))
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

chat_history = []

user_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{user_input}")
])

while True:
    question = input("Ask a question (or type 'exit' to quit): ")

    if question.lower() == "exit":
        print("Exiting the chatbot. Goodbye!")
        break

    user_prompt = user_prompt_template.format_messages(
        chat_history=chat_history,
        user_input=question
    )

    response = llm.invoke(user_prompt)

    print("AI Response:", response.content)

    # Save the conversation
    chat_history.append(HumanMessage(content=question))
    chat_history.append(AIMessage(content=response.content))

    print(chat_history)

```
