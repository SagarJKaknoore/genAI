```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2:latest")
print("Model is Ready")
```

```python
from langchain_core.prompts import ChatPromptTemplate

# Create Chat Prompt Template
user_chat_prompt_template = ChatPromptTemplate([
    ("user", "What is the capital city of {country}?")
])

# Provide value for the variable
user_prompt = user_chat_prompt_template.invoke({
    "country": "France"
})

# Send the formatted chat prompt to the LLM
response = llm.invoke(user_prompt)

# Print response
print(type(response))
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate

# Create Chat Prompt Template
user_chat_prompt_template = ChatPromptTemplate([
    ("user", "Please explain the {concept} in simple terms."),
    ("system", "You are illiterate Person who cannot read or write, Please explain the {concept} in simple terms that I can understand.")
])

# Provide value for the variable
user_prompt = user_chat_prompt_template.invoke({
    "concept": "pythagorean theorem"
})

# Send the formatted chat prompt to the LLM
response = llm.invoke(user_prompt)

# Print response
print(type(response))
print(response)
```
