```python
#Example of Knowledgebase in text format
from langchain_community.document_loaders import TextLoader
loader=TextLoader('genAi.txt')
text_pages = loader.load()
print(f'No of pages: {len(text_pages)}')
```

```python
# Example - Knowledgebase in csv format
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("employees.csv")
csv_pages = loader.load()
print(f'No of pages: {len(csv_pages)}')
```

```python
# Example - Knowledgebase in web document format
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader("https://en.wikipedia.org/wiki/Generative_AI")
web_pages = loader.load()
print(f'No of pages: {len(web_pages)}')
```
