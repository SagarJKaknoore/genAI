```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2:latest");
print("Model Ready")
```

```python
from langchain_core.prompts import ChatPromptTemplate
user_prompt_template = ChatPromptTemplate([("user", "Whos is famous indian {game} player ?")])
user_prompt = user_prompt_template.format(game="cricket")
response = model.invoke(user_prompt)
print(type(response))
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate
user_prompt_template = ChatPromptTemplate([("user", "What is {topic} ? in simple terms"), ("system", "you are a {topic} expert")])
user_prompt = user_prompt_template.format(topic="AI")
response = model.invoke(user_prompt)
print(type(response))
print(response)
```
