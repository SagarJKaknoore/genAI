```python
from langchain_ollama import ChatOllama
model = ChatOllama(model='gpt-oss:120b-cloud')
print("Model Is Ready")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field


class Person(BaseModel):
    name: str = Field(description="Person's name")
    dob: str = Field(description="Person's date of birth")
    hobbies: list[str] = Field(description="Person's hobbies")


json_outputParser = JsonOutputParser(
    pydantic_object=Person
)


prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    (
        "user",
        "#Format: {format_instructions}\n\n"
        "Question: {question}"
    )
]).partial(
    format_instructions=json_outputParser.get_format_instructions()
)


chain = prompt | model | json_outputParser


response = chain.invoke({
    "question":
    "Ivan is born 20-Sep-2000 and his hobbies are Music, cricket and football."
})


print(response)
```
